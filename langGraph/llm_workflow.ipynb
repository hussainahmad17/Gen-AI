```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict
```

```python
# define the state
class LLMState(TypedDict):
    question: str
    answer: str
```

```python
def llm_calling(state: LLMState) -> LLMState:
    model = ChatOpenAI(model_name="gpt-4", temperature=0.7)
    response = model.invoke(state['question'])
    state['answer'] = response.content
    return state
```

```python
# define the graph

graph = StateGraph(LLMState)

# define the nodes

graph.add_node("llm_calling", llm_calling)

# define the edges

graph.add_edge(START, "llm_calling")
graph.add_edge("llm_calling", END)

# compile the graph

workflow = graph.compile()
```

```python
# execute the workflow

result = workflow.invoke({"question": "What is the capital of France?"})
print(result)
```
